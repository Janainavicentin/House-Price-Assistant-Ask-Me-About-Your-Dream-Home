```python
import pandas as pd

df = pd.read_csv("Desktop/land-registry-house-prices-borough.csv", encoding="latin1")

df.info()

df.head()

print(df["Area"].unique())

print(df["Year"].unique())
```

```python
borough = "Camden"
year = "2012"

filtered = df[
    (df["Area"] == borough) &
    (df["Year"].astype(str).str.contains(year, case=False, na=False))
]
```

```python
if not filtered.empty:
    raw_value = filtered.iloc[0]["Value"]
    value = float(str(raw_value).replace(",", ""))
    print(f"💷 The median house price in {borough} in {year} was £{value:,.0f}.")
else:
    print("No matching data found.")
```

```python
import string  # Put this at the top of your notebook

def get_house_price(query):
    boroughs = df["Area"].unique()
    borough = None
    year = None

    print("🔎 User query:", query)

    # Step 1: Match the borough name in the user query
    for b in boroughs:
        if b.lower() in query.lower():
            borough = b
            print(f"✅ Borough matched: {borough}")
            break

    # Step 2: Extract a 4-digit year from the query
    for word in query.split():
        cleaned = word.strip(string.punctuation)
        if cleaned.isdigit() and len(cleaned) == 4:
            year = cleaned
            print(f"✅ Year matched: {year}")
            break

    # Step 3: Search the DataFrame using those values
    if borough and year:
        filtered = df[
            (df["Area"] == borough) &
            (df["Year"].astype(str).str.contains(year, case=False, na=False))
        ]

        if not filtered.empty:
            raw_value = filtered.iloc[0]["Value"]
            try:
                value = float(str(raw_value).replace(",", ""))
                return f"💷 The median house price in {borough} in {year} was £{value:,.0f}."
            except:
                return f"⚠️ Data found but could not convert price for {borough} in {year}."
        else:
            return f"😕 No data found for {borough} in {year}."
    else:
        return "❌ Sorry, I couldn't understand the borough or year from your question."


print(get_house_price("What was the median house price in Camden in 2012?"))
print(get_house_price("Tell me how much houses cost in Hackney in 2015"))
print(get_house_price("Was Barking and Dagenham expensive in 2018?"))
print(get_house_price("How much was the average price in Westminster in 2016?"))
```

```python
import gradio as gr

def chat_price(question):
    return get_house_price(question)

gr.Interface(
    fn=chat_price,
    inputs=gr.Textbox(
        label="💭 Ask me about your dream house",
        placeholder="Try: What was the average house price in Hackney in 2015?"
    ),
    outputs="text",
    title="🏠 House Price Assistant"
).launch()
```

```python
from gtts import gTTS
import os

def speak_house_price(query):
    answer = get_house_price(query)
    print(answer)

    tts = gTTS(answer, lang='en')
    tts.save("output.mp3")

    # On Windows:
    os.system("start output.mp3")

    # On Mac:
    # os.system("afplay output.mp3")
```
